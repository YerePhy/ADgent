```python
import json
from pyprojroot import here

# Load the registry
registry_path = here("registry.json")
with open(registry_path) as f:
    registry = json.load(f)

# Get the centaur paper entry
centaur = registry["centaur"]
paper_info = centaur["papers"][0]

paper_path = here("data") / paper_info["path"]
print(f"Title: {paper_info['title']}")
print(f"Authors: {paper_info['authors']}")
print(f"Path: {paper_path}")
print(f"Exists: {paper_path.exists()}")
```

```python
from docling.document_converter import DocumentConverter

converter = DocumentConverter()
result = converter.convert(str(paper_path))

md_content = result.document.export_to_markdown()
print(md_content[:3000])
```

```python
for i, table in enumerate(result.document.tables):
    print(f"--- Table {i} ---")
    display(table.export_to_dataframe(doc=result.document))
```
